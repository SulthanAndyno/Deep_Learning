```python
import numpy as np

a = np.array([1,2,3,4,5])

print(a)
```

```python
import time

a = np.random.rand(1000000)
b = np.random.rand(1000000)

awal = time.time()
c = np.dot(a,b)
akhir = time.time()

print(c)
print("Versi Vektor:" + str(1000*(akhir-awal)) + " ms")

c = 0
awal = time.time()
for i in range(1000000):
    c += a[i] * b[i]
akhir = time.time()

print(c)
print("Versi For Loop: " + str(1000*(akhir-awal)) + " ms")
```
